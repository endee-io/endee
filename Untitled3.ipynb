```python
import face_recognition
import os

known_encodings = []
known_names = []

for img_name in os.listdir("data/employees"):
    img_path = f"data/employees/{img_name}"
    img = face_recognition.load_image_file(img_path)
    enc = face_recognition.face_encodings(img)

    if len(enc) > 0:
        known_encodings.append(enc[0])
        known_names.append(os.path.splitext(img_name)[0])

print("Faces encoded:", known_names)
```

```python
import numpy as np
import cv2
from datetime import datetime

img = face_recognition.load_image_file(test_image_path)
encodings = face_recognition.face_encodings(img)

results = []

for enc in encodings:
    distances = face_recognition.face_distance(known_encodings, enc)
    best = np.argmin(distances)
    confidence = 1 - distances[best]

    if distances[best] < 0.6:
        name = known_names[best]
    else:
        name = "Unknown"

        # Save unknown face automatically
        os.makedirs("unknown_faces", exist_ok=True)
        cv2.imwrite(f"unknown_faces/unknown_{datetime.now().strftime('%H%M%S')}.jpg", cv2.imread(test_image_path))

    results.append((name, round(confidence*100,2)))

for r in results:
    print(f"Detected: {r[0]}, Confidence: {r[1]}%")
```

```python
import pandas as pd

file_name = "Attendance.csv"
now = datetime.now()

if os.path.exists(file_name):
    df = pd.read_csv(file_name)
else:
    df = pd.DataFrame(columns=["Name","Date","Time"])

today = str(now.date())

for r in results:
    name = r[0]
    if name == "Unknown":
        print("❌ Face not recognized")
        continue

    if ((df["Name"] == name) & (df["Date"] == today)).any():
        print(f"⚠️ {name} already marked today")
    else:
        new = {"Name": name, "Date": today, "Time": now.strftime("%H:%M:%S")}
        df = pd.concat([df, pd.DataFrame([new])], ignore_index=True)
        df.to_csv(file_name, index=False)
        print(f"✅ Attendance marked for {name}")

print("\n📊 Attendance History:")
print(df)
```

```python
import face_recognition
import numpy as np
import cv2
import os
from datetime import datetime
import pandas as pd

# --- Upload / set your test image ---
# test_image_path = "unknown.jpg"  # Already uploaded using files.upload()
print("Testing image:", test_image_path)

# --- Load test image ---
img = face_recognition.load_image_file(test_image_path)
encodings = face_recognition.face_encodings(img)

# --- Check if any face detected ---
if len(encodings) == 0:
    print("⚠️ No face detected in the uploaded image")
else:
    results = []

    for enc in encodings:
        distances = face_recognition.face_distance(known_encodings, enc)
        best = np.argmin(distances)
        confidence = 1 - distances[best]

        if distances[best] < 0.6:
            name = known_names[best]  # Known employee
        else:
            name = "Unknown"
            # Save unknown face for reference
            os.makedirs("unknown_faces", exist_ok=True)
            img_bgr = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
            cv2.imwrite(f"unknown_faces/unknown_{datetime.now().strftime('%H%M%S')}.jpg", img_bgr)

        results.append((name, round(confidence*100,2)))

    # --- Show recognition results ---
    for r in results:
        print(f"Detected: {r[0]}, Confidence: {r[1]}%")

    # --- Attendance Logic ---
    file_name = "Attendance.csv"
    now = datetime.now()
    today = str(now.date())

    if os.path.exists(file_name):
        df = pd.read_csv(file_name)
    else:
        df = pd.DataFrame(columns=["Name","Date","Time"])

    for r in results:
        name = r[0]
        if name == "Unknown":
            print("\n❌ Face not recognized")
            # Empty attendance for unknown
            empty_df = pd.DataFrame(columns=["Name","Date","Time"])
            print("\n📊 Attendance History:")
            print(empty_df)
        else:
            # Prevent duplicate marking
            if ((df["Name"] == name) & (df["Date"] == today)).any():
                print(f"⚠️ {name} already marked today")
            else:
                new = {"Name": name, "Date": today, "Time": now.strftime("%H:%M:%S")}
                df = pd.concat([df, pd.DataFrame([new])], ignore_index=True)
                df.to_csv(file_name, index=False)
                print(f"✅ Attendance marked for {name}")

    # --- Show full attendance table ---
    if any(r[0] != "Unknown" for r in results):
        print("\n📊 Full Attendance History:")
        print(df)
```

```python
import pandas as pd
from datetime import datetime

# -----------------------------
# SETTINGS
# -----------------------------
CONF_THRESHOLD = 60
file_name = "attendance.csv"

# -----------------------------
# TEMP INPUT (REMOVE LATER)
# -----------------------------
name = "Unknown"       # change to "boli photo" / "poojitha photo"
confidence = 42.1

today = datetime.now().strftime("%Y-%m-%d")
now = datetime.now()

# -----------------------------
# LOAD CSV
# -----------------------------
try:
    df = pd.read_csv(file_name)
except:
    df = pd.DataFrame(columns=["Name", "Date", "Time"])

# -----------------------------
# FORCE UNKNOWN
# -----------------------------
if confidence < CONF_THRESHOLD:
    name = "Unknown"

print(f"Detected: {name}, Confidence: {confidence:.1f}%")

# -----------------------------
# UNKNOWN CASE
# -----------------------------
if name == "Unknown":
    print("❌ Face not recognized")

    empty_df = pd.DataFrame(columns=["Name", "Date", "Time"])
    print("\n📊 Attendance History:")
    print(empty_df)

# -----------------------------
# KNOWN CASE
# -----------------------------
else:
    if ((df["Name"] == name) & (df["Date"] == today)).any():
        print(f"⚠️ {name} already marked today")
    else:
        new = {
            "Name": name,
            "Date": today,
            "Time": now.strftime("%H:%M:%S")
        }
        df = pd.concat([df, pd.DataFrame([new])], ignore_index=True)
        df.to_csv(file_name, index=False)
        print(f"✅ Attendance marked for {name}")

    print("\n📊 Full Attendance History:")
    print(df)
```
