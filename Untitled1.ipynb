```python
import face_recognition
import os

known_encodings = []
known_names = []

for img_name in os.listdir("data/employees"):
    img_path = f"data/employees/{img_name}"
    img = face_recognition.load_image_file(img_path)

    encodings = face_recognition.face_encodings(img)

    if len(encodings) == 0:
        print(f"❌ No face in {img_name}")
        continue

    known_encodings.append(encodings[0])
    name = os.path.splitext(img_name)[0]
    known_names.append(name)

    print(f"✅ Encoded: {name}")

print("🎉 All faces encoded")
```

```python
import face_recognition
import numpy as np

img = face_recognition.load_image_file(photo)
test_encoding = face_recognition.face_encodings(img)[0]

matches = face_recognition.compare_faces(known_encodings, test_encoding)
face_distances = face_recognition.face_distance(known_encodings, test_encoding)

best_match_index = np.argmin(face_distances)

if matches[best_match_index]:
    name = known_names[best_match_index]
else:
    name = "Unknown"

print(f"✅ Detected: {name}")
```
